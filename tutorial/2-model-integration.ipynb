## Model Integration with Google Gemini and GROQ

```python
### Importing the required libraries

from dotenv import load_dotenv
load_dotenv()
```

### Google Gemini model integration

```python
### Using init_chat_model to initialize the chat model with Google Gemini

from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-3.1-flash-lite", timeout=20, max_retries=0)
response = model.invoke("Why do parrots talk?")
response
```

```python
### Using ChatGoogleGenerativeAI to initialize the chat model with Google Gemini

from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", timeout=20, max_retries=0)
response = model.invoke("Why do parrots talk?")
response
```

### GROQ model integration

```python
### Using init_chat_model to initialize the chat model with GROQ

from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3.8-27b")
response = model.invoke("Why do parrots talk?")
response
```

```python
# Using ChatGROQ to initialize the chat model with GROQ

from langchain_groq import ChatGroq

model = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0.3,
    max_tokens=150,
    timeout=30,
    max_retries=2
)
response = model.invoke("Why do parrots talk?")
print(response.content)
```

### Streaming and Batch

#### Streaming

Most models can stream their responses. This means that you can get the response as it is being generated, rather than waiting for the entire response to be generated before receiving it. This can be useful for applications where you want to display the response to the user as it is being generated, or for applications where you want to process the response as it is being generated. Calling stream() on a model will return a generator that yields the response as it is being generated. You can then iterate over the generator to get the response as it is being generated.

```python
question = "Why do parrots have colorfull feathers?"

for chunk in model.stream(question):
    print(chunk.text, end="", flush=True)
```

#### Batch

Some models can also process multiple requests in a single batch. This means that you can send multiple requests to the model at once, and the model will process them all in a single batch. This can be useful for applications where you want to process multiple requests at once, or for applications where you want to reduce the number of requests you are sending to the model. Calling batch() on a model will return a generator that yields the responses for each request in the batch. You can then iterate over the generator to get the responses for each request in the batch.

```python
responses = model.batch([
    "How do airplane fly in 50 words?",
    "How do birds fly?",
    "How do fish swim?",],
    config= {
        'max_concurrency': 2, # Limit to 2 parallel requests
    })

for response in responses:
    print(response.content)
```
